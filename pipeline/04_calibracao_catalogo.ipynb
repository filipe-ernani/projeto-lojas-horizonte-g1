```sql
-- Databricks SQL
SELECT categoria,
       COUNT(*)                            AS itens,
       ROUND(percentile_approx(preco, 0.25), 2) AS p25,
       ROUND(percentile_approx(preco, 0.50), 2) AS mediana,
       ROUND(percentile_approx(preco, 0.75), 2) AS p75
FROM horizonte.silver.marketplace_itens
WHERE valido AND incluir AND periodo_completo
GROUP BY categoria
ORDER BY itens DESC;
```
