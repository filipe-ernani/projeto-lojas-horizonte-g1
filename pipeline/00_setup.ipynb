# 00 · Setup do ambiente
Cria a estrutura medalhão do projeto Lojas Horizonte no Unity Catalog.

| Databricks (MVP) | Equivalente no GCP |
|---|---|
| Catálogo `horizonte` | Projeto `lojas-horizonte` |
| Schemas `bronze`, `silver`, `gold` | Datasets do BigQuery |
| Volume `bronze.landing` | Bucket de landing no Cloud Storage |

Rodar uma única vez. Os comandos usam `IF NOT EXISTS`, então podem ser executados de novo sem efeito colateral.


### Criação do catálogo, schemas e volume
- **bronze:** dados como chegaram das fontes, sem tratamento
- **silver:** dados limpos, tipados e com regras de negócio aplicadas
- **gold:** tabelas agregadas, prontas para o dashboard e para a IA
- **landing:** volume onde os arquivos brutos (CSV) são depositados antes da ingestão

```sql
CREATE CATALOG IF NOT EXISTS horizonte;
CREATE SCHEMA IF NOT EXISTS horizonte.bronze;
CREATE SCHEMA IF NOT EXISTS horizonte.silver;
CREATE SCHEMA IF NOT EXISTS horizonte.gold;
CREATE VOLUME IF NOT EXISTS horizonte.bronze.landing;
```
